1.  Get user input using **`input(“Enter your age: ”)`**. If user is 18 or older, give feedback: **`You are old enough to drive`**. If below 18 give feedback to **`You need #number more years to learn to drive`**.
​
Output:
​
```py
    Enter your age: 30
    You are old enough to learn to drive.

```

2.  Compare the values of my_age and your_age using **`if-else`**. Who is older (me or you)? Use **`input(“Enter your age: ”)`** to get the age as input. You can use a nested condition to print **`'year'`** for 1 year difference in age, **`'years'`** for bigger differences, and a custom text if **`my_age = your_age`**.

​
Output:
​
```py
    Enter your age: 30
    You are 5 years older than me.
```

````python
MY_AGE = 30
your_age = int(input('Enter your age: '))
unit = ['\b' + 'one year', ' years']
difference = MY_AGE - your_age
if difference > 0:
  print(f'I\'m {difference}{unit[min(1, difference - 1)]} older than you.')
elif difference < 0:
  difference = abs(difference)
  print(f'you are {difference}{unit[min(1, difference - 1)]} older than me.')
else:
  print('You are the same age.')
````

3. Check if the season is Autumn, Winter, Spring or Summer.
    - If the user input is:
        - September, October or November, the season is Autumn.
        - December, January or February, the season is Winter.
        - March, April or May, the season is Spring
        - June, July or August, the season is Summer


````python
seasons = {
    "Autumn": ["september", "october", "november"],
    "Winter": ["december", "january", "february"],
    "Spring": ["march", "april", "may"],
    "Summer": ["june", "july", "august"]
}
month = input('Enter a month: ').lower()
for season in seasons:
  if month in seasons[season]:
    print(f'The season is {season}')
    break
````

# Part 2: Dictionary & Conditions Practice

You are given a dictionary called `person` that contains information about an individual.  
Your task is to perform the following operations step-by-step:

##  The Dictionary:

```python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
```

## Tasks

### 1. Check for `skills` key
- If the `skills` key exists in the dictionary:
  - Print the **middle skill** from the list of skills.

### 2. Check for `'Python'` in skills
- Still inside the same condition:
  - Check if `'Python'` is in the person's skills list.
  - Print `True` if found, otherwise `False`.

### 3. Based on the combination of skills:
  - If the person has **only** `['Python', 'Matlab']` → print: `He knows machine learning`
  - If the person has both `'Python'` and `'R'` → print: `He knows statistics`
  - If the person has both `'C'` and `'C++'` → print: `He knows software development`
  - Else → print: `unknown title`


### 4. Print location and marital status
- If the person is married **and** lives in **Finland**, print:
```py
    Milaan Parmar lives in Finland. He is married.
```



````python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
````

````python
#1
if 'skills' in person.keys():
  print(f'The middle skill: {person['skills'][len(person['skills']) // 2]}')
````

````python
#2
skill_to_check = 'python'
print(f'Python existence in person skills: {skill_to_check in list(map(str.lower, person['skills']))}')
````

````python
#3
titles = {
    ('Python', 'Matlab'): 'He knows machine learning',
    ('Python', 'R'): 'He knows statistics',
    ('C', 'C++'): 'He knows software development'
}
flag = True
for title in titles:
  for x in title:
   if x not in person['skills']:
    flag = False
    break
  if flag:
    print(titles[title])
````

````python
#4
country = 'FINLAND'
if person['country'].upper() == country and person['is_marred']:
  print(f'{person['first_name']} {person['last_name']} lives in {country}. He is married')
````

   1. What are the total number of languages in the data ( output:  112 )
   2. Find the ten most spoken languages from the data ( output:  English , French , Arabic )

````python
from countries import countriesWWWWW
from countries_details_dat import countries_data
````

````python
# 1
counntries_land = [x for x in countriesWWWWW if 'land' in x]
print(counntries_land)
````

````python
# 2.1
languages = []
for country in countries_data:
  languages.extend(country['languages'])
print(len(set(languages)))
````

````python
#2.2
from collections import Counter
top_10 = [x[0] for x in Counter(languages).most_common(10)]
print('output: ', ', '.join(top_10))
````

````python
#2.3
top_10_population= sorted(countries_data, key=lambda x: x['population'], reverse=True)[:10]
top_10_population_names = [x['name'] for x in top_10_population]
print('output: ', ', '.join(top_10_population_names))
````
