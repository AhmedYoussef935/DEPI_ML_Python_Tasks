````python
# 1
import math
def area_of_circle(radius):
    return math.pi * radius ** 2
def perimeter_of_cirlce(radius):
    return 2 * math.pi * radius
# 2
def sum_all_nums(*args):
    if not all(isinstance(i, (int, float)) for i in args):
        raise Exception('Enter Numbers Only')
    return sum(args)
# 3
def convert_celsius_2_fahrenheit(T_in_C):
    return T_in_C * (9 / 5) + 32
# 4
def check_season(month):
    SEASONS = {
    "Autumn": ["september", "october", "november"],
    "Winter": ["december", "january", "february"],
    "Spring": ["march", "april", "may"],
    "Summer": ["june", "july", "august"]
    }
    for season in SEASONS:
        if month in SEASONS[season]:
            return season
# 5
def calculate_slope(p1: tuple, p2: tuple):
    try:
        return (p2[1] - p1[1]) / (p2[0] - p1[0])
    except ZeroDivisionError as ex:
        raise Exception('Error! Divide By Zero') from ex
# 6
def solve_quadratic_eqn(a, b, c):
    #solution law: x = -b +-sqrt(b^2 - 4ac) / 2a
    if a == 0:
        raise Exception('The equation is not a Q uadratic Equation')
    delta = b ** 2 - 4 * a * c
    deno = 2 * a
    if delta >= 0:
        x1, x2 = (-b + delta ** 0.5) / deno, (-b - delta ** 0.5) / deno
    else:
        x1, x2 = complex(-b / deno, abs(delta) ** 0.5 / deno), complex(-b / deno, -1 * abs(delta) ** 0.5 / deno)
    return x1 if x1 == x2 else (x1, x2)
# 7
def print_list(names):
    print('Output: ', ', '.join(names))
# 8
def capitalize_list_items(items):
    return list(map(str.capitalize, items))
# 9
def add(l: list, item):
    l.append(item)
    return l
# 10
def remove_item(l: list, item):
    l.remove(item)
    return l
# 11
def sum_of_numbers(num):
    return sum(range(num + 1))
# 12
def sum_of_odd(num):
    return sum(range(1, num + 1, 2))
# 13
def sum_of_even(num):
    return sum(range(0, num + 1, 2))
````

````python
# 1
print("1.")
print("Area:", area_of_circle(5))
print("Perimeter:", perimeter_of_cirlce(5))


# 2
print("\n2.")
print(sum_all_nums(1, 2, 3, 4, 5))
print(sum_all_nums(10, 2.5, 7))


# 3
print("\n3.")
print(convert_celsius_2_fahrenheit(0))
print(convert_celsius_2_fahrenheit(100))


# 4
print("\n4.")
print(check_season("january"))
print(check_season("april"))
print(check_season("july"))
print(check_season("october"))


# 5
print("\n5.")
print(calculate_slope((1, 2), (3, 6)))
print(calculate_slope((2, 5), (4, 9)))


# 6
print("\n6.")
print(solve_quadratic_eqn(1, -5, 6))   # two real roots
print(solve_quadratic_eqn(1, -4, 4))   # one real root
print(solve_quadratic_eqn(1, 4, 7))    # two complex roots


# 7
print("\n7.")
print_list(["Ahmed", "Ali", "Omar"])


# 8
print("\n8.")
print(capitalize_list_items(["hello", "python", "world"]))


# 9
print("\n9.")
my_list = [1, 2, 3]
print(add(my_list, 4))


# 10
print("\n10.")
my_list = [1, 2, 3, 4]
print(remove_item(my_list, 3))


# 11
print("\n11.")
print(sum_of_numbers(5))
print(sum_of_numbers(10))


# 12
print("\n12.")
print(sum_of_odd(10))
print(sum_of_odd(7))


# 13
print("\n13.")
print(sum_of_even(10))
print(sum_of_even(7))
````

5. Write different functions which take lists. They should **`calculate_mean`**, **`calculate_median`**, **`calculate_mode`**
6. Declare a function named **`print_list`**. It takes a list as a parameter and it prints out each element of the list.

````python
from collections import Counter
# 1
def capitalize_list_items(items):
    return list(map(str.capitalize, items))
# 2
def add(l: list, item):
    l.append(item)
    return l
# 3
def reverse_list(l: list):
    reversed_list = [l[-i] for i in range(1, len(l) + 1)]
    return reversed_list
# 4
def evens_and_odds(num):
    evens = num // 2 + 1
    odds = num // 2
    if num % 2 != 0:
        odds += 1
    return {'evens': evens, 'odds': odds}
# 5.1
def calculate_mean(l):
    return sum(l) / len(l)
# 5.2
def calculate_median(l):
    sorted_l = sorted(l)
    if len(l) % 2 == 0:
        median = (sorted_l[int(len(l) / 2 - 1)] + sorted_l[int(len(l) / 2)]) / 2
    else:
        median = sorted_l[len(l) // 2]
    return median
# 5.3
def calculate_mode(l):
    count = Counter(l)
    max_count = count.most_common(1)[0][1]
    output = []
    for x in count:
        if count[x] == max_count:
            output.append(x)
    return output if len(output) > 1 else output[0]
# 6
def print_list(items):
    print('List Items: ', ', '.join(items))
````

````python
# 1
print("1.")
print(capitalize_list_items(["hello", "PYTHON", "world"]))


# 2
print("\n2.")
my_list = [1, 2, 3]
print(add(my_list, 4))


# 3
print("\n3.")
print(reverse_list([1, 2, 3, 4, 5]))


# 4
print("\n4.")
print(evens_and_odds(5))
print(evens_and_odds(6))


# 5.1
print("\n5.1.")
print(calculate_mean([1, 2, 3, 4, 5]))


# 5.2
print("\n5.2.")
print(calculate_median([1, 2, 3, 4, 5]))
print(calculate_median([1, 2, 3, 4]))


# 5.3
print("\n5.3.")
print(calculate_mode([1, 2, 2, 3, 4]))
print(calculate_mode([1, 2, 2, 3, 3, 4]))


# 6
print("\n6.")
print_list(["Ahmed", "Ali", "Omar"])
````

### Exercises ➞ <span class='label label-default'>Level 3</span>

1. Write a function called **`is_prime`**, which checks if a number is prime.
2. Write a functions which checks if all items are unique in the list.
3. Write a function which checks if all the items of the list are of the same data type.
4. Write a function which check if provided variable is a valid python variable

````python
# 1
def is_prime(num):
    for x in range(2, int(num ** 0.5) + 1):
        if num % x == 0:
            return False
    return True and num > 1
# 2 
def is_unique(l):
    return len(l) == len(set(l))
# 3
def is_same_type(l):
    unique_types = set(map(type, l))
    return len(unique_types) == 1
# 4
def check_var_name(var_name):
    Keywords = [
    'False', 'None', 'True',
    'and', 'as', 'assert', 'async', 'await',
    'break', 'case', 'class', 'continue',
    'def', 'del', 'elif', 'else',
    'except', 'finally', 'for', 'from',
    'global', 'if', 'import', 'in', 'is',
    'lambda', 'match', 'nonlocal', 'not',
    'or', 'pass', 'raise', 'return',
    'try', 'while', 'with', 'yield'
    ]
    l = list(range(48, 58)) + list(range(65, 91)) + [95]
    if var_name in Keywords:
        return False
    var_name = var_name.upper()
    for i in range(len(var_name)):
        if i == 0 and ord(var_name[i]) in range(48, 58):
            return False
        if not ord(var_name[i]) in l:
            return False
    return True
# Using Buit-in function (iskeyword - isidentifier)
import keyword
def check_var_nameV02(var_name: str):
    return var_name.isidentifier() and not keyword.iskeyword(var_name)
````
